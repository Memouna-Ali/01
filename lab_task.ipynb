Take the name, roll number and marks (Python, AI, Mathematics) of 5 students. Store each student as a dictionary in a list. Use a function to calculate total, percentage and grade (A: 80–100, B: 70–79, C: 60–69, D: 50–59, F: below 50). Display all results, show the student with the highest percentage, and save the results to results.txt.

```python
def calculate_result(python, ai, maths):
    total = python + ai + maths
    percentage = total / 300 * 100
    if percentage >= 80:
        grade = "A"
    elif percentage >= 70:
        grade = "B"
    elif percentage >= 60:
        grade = "C"
    elif percentage >= 50:
        grade = "D"
    else:
        grade = "F"
    return total, percentage, grade


students = []

for i in range(5):
    print("Student", i + 1)
    name = input("Enter name: ")
    roll = input("Enter roll number: ")
    python = int(input("Python marks: "))
    ai = int(input("Artificial Intelligence marks: "))
    maths = int(input("Mathematics marks: "))

    total, percentage, grade = calculate_result(python, ai, maths)

    student = {"name": name,"roll": roll,"total": total,"percentage": percentage,"grade": grade}
    students.append(student)

file = open("results.txt", "w")

print("\n--- RESULTS ---")
file.write("--- RESULTS ---\n")

for s in students:
    line = (s["name"] + " | Roll: " + s["roll"] + " | Total: " + str(s["total"])
            + " | Percentage: " + str(s["percentage"]) + " | Grade: " + s["grade"])
    print(line)
    file.write(line + "\n")

top = students[0]
for s in students:
    if s["percentage"] > top["percentage"]:
      top = s

print("\nHighest percentage:", top["name"], "-", top["percentage"])
file.write("\nHighest percentage: " + top["name"] + " - " + str(top["percentage"]) + "\n")

file.close()
```

Take a sentence from the user. Count the words, characters and vowels, find the longest and shortest word, create a set of unique words, and display them in alphabetical order. Use strings, lists/sets and functions. Save the analysis to word_analysis.txt.

```python
sentence=input("Enter a sentence: ")
len(sentence)
words=sentence.split()
print("Number of characters in the sentence:",len(sentence))#total characters
print("Number of words in the sentence:",len(words))#total words
#total vowels
def count_vowels(sentence):
    vowels = "aeiouAEIOU"
    count = 0
    for char in sentence:
        if char in vowels:
            count += 1
    return count
print("Number of vowels in the sentence:",count_vowels(sentence))
def longest_word(sentence):
    words = sentence.split()
    longest = ""
    for word in words:
        if len(word) > len(longest):
            longest = word
    return longest
print("Longest word in the sentence:",longest_word(sentence))
def shortest_word(sentence):
    words = sentence.split()
    shortest = words[0]
    for word in words:
        if len(word) < len(shortest):
            shortest = word
    return shortest
print("Shortest word in the sentence:",shortest_word(sentence))
def alphabetic_order(sentence):
    words = sentence.split()
    words.sort()
    return " ".join(words)
print("Sentence in alphabetical order:",alphabetic_order(sentence))
file=open("word_analysis.txt","w")
file.write("Number of characters in the sentence: " + str(len(sentence)) + "\n")
file.write("Number of words in the sentence: " + str(len(words)) + "\n")
file.write("Number of vowels in the sentence: " + str(count_vowels(sentence)) + "\n")
file.write("Longest word in the sentence: " + longest_word(sentence) + "\n")
file.write("Shortest word in the sentence: " + shortest_word(sentence) + "\n")
file.write("Sentence in alphabetical order: " + alphabetic_order(sentence) + "\n")
file.close()
```


Take the name, price and quantity of 5 products and store them in dictionaries. Calculate each product's total and the complete bill. Apply a 10% discount if the bill is over Rs. 10,000. Display the final bill and the most expensive product, and save the bill to bill.txt. Use a function to calculate the bill.

```python
def calculate_bill(products):
    bill = 0
    for p in products:
        bill = bill + p["total"]
    if bill > 10000:
        bill = bill - bill * 0.10
    return bill


products = []

for i in range(5):
    print("Product", i + 1)
    name = input("Product name: ")
    price = float(input("Price: "))
    quantity = int(input("Quantity: "))

    product = {"name": name,"price": price,"quantity": quantity, "total": price * quantity}
    products.append(product)

complete_bill = 0
for p in products:
    complete_bill = complete_bill + p["total"]

final_bill = calculate_bill(products)

expensive = products[0]
for p in products:
    if p["price"] > expensive["price"]:
        expensive = p

print("\nTotal for each product:")
for p in products:
    print(p["name"], ":", p["total"])

print("Complete bill:", complete_bill)
print("Final bill:", final_bill)
print("Most expensive product:", expensive["name"])

file = open("bill.txt", "w")
for p in products:
  file.write(p["name"] + " : " + str(p["total"]) + "\n")
file.write("Complete bill: " + str(complete_bill) + "\n")
file.write("Final bill: " + str(final_bill) + "\n")
file.write("Most expensive product: " + expensive["name"] + "\n")
file.close()
```

Take two numbers and let the user choose an operation (+, −, *, /, %). Perform it using if/elif/else, display the result, and ask whether to do another calculation. Repeat until the user exits. Use a separate function for each operation.


```python
def add(a, b):
    return a + b


def subtract(a, b):
    return a - b


def multiply(a, b):
    return a * b


def divide(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a / b


def modulus(a, b):
    if b == 0:
        return "Cannot divide by zero"
    return a % b


again = "yes"

while again == "yes":
    a = float(input("Enter first number: "))
    b = float(input("Enter second number: "))

    print("Select operation: + - * / %")
    op = input("Operation: ")

    if op == "+":
        print("Result:", add(a, b))
    elif op == "-":
        print("Result:", subtract(a, b))
    elif op == "*":
        print("Result:", multiply(a, b))
    elif op == "/":
        print("Result:", divide(a, b))
    elif op == "%":
        print("Result:", modulus(a, b))
    else:
        print("Invalid operation")

    again = input("Another calculation? (yes/no): ")

print("Program terminated")
```

Display a menu (Burger Rs. 500, Pizza Rs. 800, Fries Rs. 250, Drink Rs. 150). Let the customer select items and quantities, allow multiple items, and calculate the price and final bill. Save the bill to orders.txt. Use dictionaries, loops, functions, conditions and file handling.

```python
menu = {1: ["Burger", 500],2: ["Pizza", 800],3: ["Fries", 250],4: ["Drink", 150]}

def show_menu():
    print("\n Menu")
    for key in menu:
        print(key, ".", menu[key][0], "Rs.", menu[key][1])


def calculate_price(price, quantity):
    return price * quantity


order = []
final_bill = 0
more = "yes"

while more == "yes":
    show_menu()
    choice = int(input("Select item number: "))

    if choice in menu:
        quantity = int(input("Quantity: "))
        name = menu[choice][0]
        price = calculate_price(menu[choice][1], quantity)
        order.append([name, quantity, price])
        final_bill = final_bill + price
    else:
        print("Invalid choice")

    more = input("Order more items? (yes/no): ")

print("\nBill")
file = open("orders.txt", "w")
file.write("Bill\n")

for item in order:
    line = item[0] + " x " + str(item[1]) + " = Rs. " + str(item[2])
    print(line)
    file.write(line + "\n")

print("Final bill: Rs.", final_bill)
file.write("Final bill: Rs. " + str(final_bill) + "\n")
file.close()
```
